# Exercise - Analyzing Data Science Job Demand

```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_india = df[df['job_country'] == 'India'].copy()
```

```python
df_india
```

```python
df_india['job_posted_month'] = df_india['job_posted_date'].dt.strftime('%B')
```

```python
df_india.head(5)
```

```python
df_india_pivot = df_india.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
```

```python
df_india_pivot.reset_index(inplace=True)

df_india_pivot['job_posted_month_no'] = pd.to_datetime(df_india_pivot['job_posted_month'], format='%B').dt.month

df_india_pivot.sort_values(by='job_posted_month_no', inplace=True)
```

```python
df_india_pivot.set_index('job_posted_month',inplace=True)

df_india_pivot.drop(columns='job_posted_month_no', inplace=True)
```

```python
df_india_pivot
```

```python
df_india_pivot.plot(kind ='line')
```

```python
top_3 = df_india['job_title_short'].value_counts().head(3)
top_3 = top_3.index.tolist()
```

```python
df_india_pivot[top_3].plot(kind='line')
plt.title('Job Demand in India by Month')
plt.xlabel('')
plt.ylabel('Number of Job Postings')
plt.legend(title='Job Title', loc='upper right')
plt.show()
```
