# What Are the Most Demanded Skills for the Top 3 Most Popular Data Roles?

## 🎯 Objective
Identify the most demanded skills for the three most popular data job roles by analyzing job postings and visualizing the results using Python.

1. **Clean the Skills Column:** Clean and convert the `job_skills` column into a list of individual skills.

### Import Libraries and Data
